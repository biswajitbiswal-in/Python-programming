2. WAP to find sum of 1 to n

```python
n = int(input("Enter the number"))
i = 0 
sum = 0 
while i<=n:
    sum +=i
    i+=1
print(sum)
```

```python
n = int(input("How many numbers you want to add"))
i = 0
sum = 0 
while i<n:
    a = int(input("enter the number"))
    sum +=a
    i += 1
print(sum)
```
