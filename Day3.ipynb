```python
price = int(input("Enter the phone price"))

if price>=10000:
    discount = price / 100 * 0.1
    price = price - discount
    print("total price: ",price)
else :
    print("No discount:",price)
    
```

4. WAP to find the greatest num among three num

```python
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))
num3 = float(input("Enter third number: "))

if (num1 >= num2) and (num1 >= num3):
    largest = num1
elif (num2 >= num1) and (num2 >= num3):
    largest = num2
else:
    largest = num3

print("The largest number is :",largest)
```

6. WAP to print your grade for the input marks

```python
mark =  int(input("enter your mark"))

if mark<=100 and mark >= 90:
    print("Grade is o")
elif mark<=70 and mark >=89:
    print("Grade is A")
elif mark<=60 and mark >= 69:
    print("Grade is c")
else:
    print("Fail")
```

7. Take the marks securd in 5 subjects and then print student will get 1st div,2nd or fail

```python
sub1 = float(input("Enter marks for Sub 1: "))
sub2 = float(input("Enter marks for Sub 2: "))
sub3 = float(input("Enter marks for Sub 3: "))
sub4 = float(input("Enter marks for Sub 4: "))
sub5 = float(input("Enter marks for Sub 5: "))


total_marks = sub1 + sub2 + sub3 + sub4 + sub5
percentage = (total_marks / 500) * 100

print("Total Marks:",total_marks)
print("Percentage:%",percentage)


if percentage >= 60:
    print("Result: 1st Div")
elif percentage >= 50:
    print("Result: 2nd Div")
elif percentage >= 40:
    print("Result:3rd Div")
else:
    print("Result:Fail")
```

8. Print the week days with approprite input num 1 to 7 where 1 sunday and 7 is saturday

```python
day = int(input("enter 1 to 7 "))

if day == 1 :
    print("sunday")
elif day == 2:
    print("Monday")
elif day == 3:
    print("tues")
elif day == 4:
    print("wed")
elif day == 5:
    print("thurs")
elif day == 6:
    print("friday")
elif day == 7 :
    print("saturday")
else:
    print("not a valid day")
```

Write a python program to implement simple atm withdrawl system accept account balance and withdrawl ammount Check whether the withdrawl ammount is valid or not whether suffiecent balance in available and whether withdrawl satisfied the condition of maintaining minimum balance ammount in the account,display the updated balance or appropriate message for that

```python
 balance = 10000
min_balance = 500.0  

withdraw_amount = float(input("Enter amount to withdraw: "))


if withdraw_amount <= 0:
    print("Invalid withdrawal amount")
else:
    
    if withdraw_amount > balance:
        print("Insufficient balance")
    else:
      
        remaining_balance = balance - withdraw_amount
        
        if remaining_balance < min_balance:
            print("Withdrawal denied! ", min_balance)
        else:
          
            balance = remaining_balance
            print("Please collect your cash.")
            print("Updated account balance:", balance)
```
