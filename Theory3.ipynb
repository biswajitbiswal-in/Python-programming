# Function

WAP Function to check the number is even or odd

```python
def is_even(i):
    if i % 2 == 0:
        print("the number is even:",i)
    else:
        print("its a odd number:",i)
num = int(input("Enter the number to check num is even or odd:"))
is_even(num)
```

```python
def is_prime(i):
    count = 0
    for j in range(1, i + 1):
        if i % j == 0:
            count += 1
    return count == 2

print(is_prime(13))
```
