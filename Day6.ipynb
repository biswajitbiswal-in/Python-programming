1. WAP PROGRAM To convert a list into a tupple.

```python
def list_t(l):
    tup = ()
    for i in l:
        tup += i
        print(tup)
list1 = [1,2]
print(list_t(list1))
```

2. second largest element in list of int without using any predefined function like short or max

```python
def second_largest(nums):
    largest = nums[0]
    second_lar = nums[0]
    for num in nums:
       
        if num > largest:
            second_lar = largest
            largest = num
           
    
        elif num > second_lar and num != largest:
            second_lar = num
        return second_lar

        
list1 = [6,9,5,89,65]
print("second largest:",second_largest(list1))
```
