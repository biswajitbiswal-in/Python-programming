1. Find the square of all the number from 1 to n

```python
num = int(input("Enter the number"))

for i in range(1,num):
    square = i * i
    print(square)
    
```

2. Print factorial fo all the numbers between any two numbers

```python
num1 = int(input("Enter the first number"))
num2 = int(input("Enter the second number"))

for i in range(num1,num2 + 1):
    fact = 1
    for num in range(1,i+1):
        fact = fact*num

    print(i,"factorial","=",fact)
```

3. Print multipication table upto 30

```python
for num in range(1,31):
    print("table of num:",num)
    
    for i in range(1,11):
        print(num , "x" , i ,"=",num*i)
   
```

4. Find the sum of the  series 1-3+6-10+15-21....n>100

```python
n = int(input())
total = 0
for i in range(1, n + 1):
    term = i * (i + 1) 
    if i % 2 == 0:
        total -= term
    else:
        total += term
print(total)
```

5. find 1+1/2i + 1/3i + 1/4i ...... 1/ni

```python
n = int(input("Enter the number"))

fact = 1
sum = 0
for i in range(1,n+1):
    fact *= i
    sum = sum + (1/fact)
print(sum)
```

6. find 1-3/2i + 6 / 3i - 10 / 4i + 15 / 5i - 21 /6i..........(20 terms)

```python
sum = 0
term = 0
fact = 1
for i in range(1,21):
     term +=i
     fact*=i

     if i% 2 == 0:
         sum = sum - term/fact
     else:
         sum = sum + term/fact
print("Sum = ",sum)
```

7. sum of digits of a number

```python
## 7 sum of digits ofa number?
num=int(input("enter the number"))
sum=0
for i in range(1,num):
    digit=num%10
    sum=sum+digit
    num=num//10

    if num==0:
        break
print(sum)
```

8. Reverse a number

```python
# reverse a number?
num=input("input the number")
print(num[::-1])

        
num = int(input("Enter the number"))
rev = 0

for i in range(1, num):
    digit = num % 10
    rev = rev * 10 + digit
    num = num // 10

    if num == 0:
        break

print("Reverse =", rev)
```

9. pallidro number -> 121,1331,111

```python
#pallindrom number?
num=input("enter number")
if num==str(num[::-1]):
    print("pallindrom number")
else:
    print("not a palindrome number")
```

10. wap to find five sets of a^2+b^2+c^2+d^2 = e^2 where a<b<c<d<e.

```python
# a^2 + b^2 + c^2 + d^2 =e^2? a<b<c<d<e
for a in range(1, 10):
    for b in range(a + 1, 10):
        for c in range(b + 1, 10):
            for d in range(c + 1, 40):
                for e in range(d + 1, 50):
                    
                    if a*a + b*b + c*c + d*d == e*e:
                        print(a, b, c, d, e)
```
