```python
class compl:
    def __init__(self, m, j):
        self.m = m
        self.j = j
    def add(self,c):
        r = self.m + c.m
        i = self.j + c.j
        print("addition :",r,"+",i,"i")
    def sub(self,c):
        r = self.m - c.m
        i = self.j - c.j
        print("addition :",r,"-",i,"i")
        
c1 = compl(1, 2)
c2 = compl(2,3)
print("Addition",f"{c1.m} + {c1.j}i")
print("substraction",f"{c1.m} - {c1.j}i")
c1.add(c2)
c1.sub(c2)
```
