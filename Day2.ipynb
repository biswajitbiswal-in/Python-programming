5. WAP to find sum of digits of a five digit number

```python
a = int(input("enter 5 digit num"))
b =  a//10000
c =  (a // 1000) % 10
d =  (a // 100) % 10
e =  (a // 10) %10 
f =  a%10
s = b+c+d+e+f
print(s)
```

7. Write a program to find a power and energy consuption on given time

```python
power = int(input())
time = int(input())
energy = power * time
print("Energy consumed:",energy)
```

8. Implement simple intrest

```python
p = float(input("Enter principal amount: "))
t = float(input("Enter time (in years): "))
r = float(input("Enter rate of interest : "))

interest = (p * t * r) / 100
print("Simple Interest:", interest)
```

9. WAP to find out the distance travelled by you if you are travelling on particular speed and particular time.

```python
speed = float(input("Enter speed (e.g., km/h): "))
time = float(input("Enter time (in hours or seconds): "))

distance = speed * time
print("Distance travelled:", distance)
```

10. Temperature Converter (Celsius to Fahrenheit) (Standard input/formula problem)
Applies formula $F = (C \times \frac{9}{5}) + 32$

```python
celsius = float(input("Enter temperature in Celsius: "))
fahrenheit = (celsius * 9 / 5) + 32
print("Temperature in Fahrenheit:", fahrenheit)
```
